```python
import json
from pathlib import Path
from collections import Counter
```

```python
# Open the processed JSON file to get extracted gts data
in_file = "gts_sqlite_processed"
in_path = Path('utils/database_sozluk/v12/processed') / f"{in_file}.json"
with open(in_path, "r", encoding="utf-8") as f:
    gts_dataset = json.load(f)
gts_dataset['dulavrat otu_1']
```

```python
# Check if there is a repeating element
all_madde = [gts_dataset[k] for k in gts_dataset.keys()]
all_madde_name = [madde['madde_name'] for madde in all_madde]
name_counts = Counter(all_madde_name)
```

```python
# Get only repeating names
repeating_names = [(name, count) for name, count in name_counts.items() if count > 1]
repating_names_sorted = sorted(repeating_names, key=lambda val:val[1], reverse=True)
print(f"Number of repeating entries: {len(repating_names_sorted)}")
```
